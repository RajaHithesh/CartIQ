# CartIQ - Load Dataset and Identify Missing Values

This notebook loads the merged Instacart 50K basket dataset and identifies missing values, duplicates and basic statistics (Module 1).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATASET_PATH = "D:/ML/CartIQ/dataset/instacart_50k_Raw.csv"
df = pd.read_csv(DATASET_PATH)
print("Dataset Shape:", df.shape)
df.head()
```

## 1. First Records

```python
df.head(10)
```

## 2. Column Names and Data Types

```python
print(df.columns.tolist())
print(df.dtypes)
```

## 3. Missing Values Per Column

```python
print(df.isnull().sum())
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

## 4. Duplicate Rows

```python
print("Duplicate rows:", df.duplicated().sum())
```

## 5. Missing Values Heatmap

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, yticklabels=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.show()
```
